# Zero Tic‑Tac‑Toe — Exact Best‑Play Policy (Minimax)
This notebook solves the game by full minimax with memoization and exports a reusable policy.

```python
import zero_ttt_core as Z
from zero_ttt_minimax import MinimaxSolver, OptimalPolicy
import zero_ttt_minimax_utils as utils
```

## Solve from the initial position

```python
solver = MinimaxSolver()
entry = solver.solve_from_initial()
print("Initial value:", entry.value, "best move:", entry.best_move)  # value ∈ {-1,0,1}
```

## Using the policy later in your code

```python
pi = OptimalPolicy(solver)
s = Z.initial_state()
mv = pi.action(s)     # (v, i)
val = pi.value(s)     # -1/0/1
```

## Save/Load Policy

```python
# # 3) Save for later
# solver.save_policy("zero_ttt_minimax_policy.npy")
#
# # 4) Load later
solver = OptimalPolicy.from_file("zero_ttt_minimax_policy.npy")
mv = solver.action(Z.initial_state())
```
